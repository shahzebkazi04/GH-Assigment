```python
students = {}
n = int(input("Enter number of students: "))

for i in range(n):
    print("\nEnter details of student", i + 1)

    roll_no = int(input("Enter Roll Number: "))
    name = input("Enter Name: ")
    branch = input("Enter Branch: ")
    marks = float(input("Enter Marks: "))

    student = (name, branch, marks)

    students[roll_no] = student

print("\n--- Add New Student ---")
roll_no = int(input("Enter Roll Number: "))
name = input("Enter Name: ")
branch = input("Enter Branch: ")
marks = float(input("Enter Marks: "))

student = (name, branch, marks)
students[roll_no] = student

print("\n--- Delete Student ---")
roll_no = int(input("Enter Roll Number to delete: "))

if roll_no in students:
    del students[roll_no]
    print("Student deleted successfully.")
else:
    print("Student not found.")

print("\n--- Update Student ---")
roll_no = int(input("Enter Roll Number to update: "))

if roll_no in students:
    name = input("Enter new Name: ")
    branch = input("Enter new Branch: ")
    marks = float(input("Enter new Marks: "))

    students[roll_no] = (name, branch, marks)
    print("Student updated successfully.")
else:
    print("Student not found.")


attributes = ["Name", "Branch", "Marks"]

print("\n--- Final Student Records ---")

for roll_no, details in students.items():
    print("Roll Number:", roll_no)
    print(attributes[0] + ":", details[0])
    print(attributes[1] + ":", details[1])
    print(attributes[2] + ":", details[2])
    print()
```
